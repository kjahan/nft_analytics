## Scrape / Set params

Set the asset Azuki, Bobu, or Moonbird

```python
SCROLL_PAUSE_TIME = 2

# asset = "bobu"
asset = "moonbirds"

current_day = datetime.now().day
current_month = datetime.now().strftime("%B").lower()

date_ = "{}_{}".format(current_month, current_day)
print("Current date: {}".format(date_))

base_path = "../../data"

if asset.lower() == "azuki":
    asset_url = "https://opensea.io/collection/azuki?tab=activity"
    asset_type = "azuki"
elif asset.lower() == "bobu":
    asset_url = "https://opensea.io/collection/bobutoken?tab=activity"
    asset_type = "bobu"
elif asset.lower() == "moonbirds":
    asset_url = "https://opensea.io/collection/proof-moonbirds?tab=activity"
    asset_type = "moonbirds"
```

```python
cnt = 1
asset_fn = "{}/{}_{}_{}.html".format(base_path, asset_type, date_, cnt)
print("Asset fn: {}".format(asset_fn))

# driver = webdriver.Chrome() # if you want to use chrome, replace Firefox() with Chrome()
driver.get(asset_url) # load the web page

src = driver.page_source # gets the html source of the page


with open(asset_fn, "w") as fp:
    fp.write(src)

cnt += 1
last_height = driver.execute_script("return document.body.scrollHeight")
print("Last height: {}".format(last_height))

while True:
    res = driver.execute_script("return document.body.scrollTop")
    print("Res before: {}".format(res))
    
    # Scroll down to bottom
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    
    res = driver.execute_script("return document.body.scrollTop")
    print("Res after: {}".format(res))

    # Wait to load page
    time.sleep(15 + randrange(5, 10))
    
    src = driver.page_source # gets the html source of the page

    ## Write to disk
    asset_fn = "{}/{}_{}_{}.html".format(base_path, asset_type, date_, cnt)
    print("Asset fn: {}".format(asset_fn))


    with open(asset_fn, "w") as fp:
        fp.write(src)

    # Wait to load page
    time.sleep(SCROLL_PAUSE_TIME)

    # Calculate new scroll height and compare with last scroll height
    new_height = driver.execute_script("return document.body.scrollHeight")
    print("New height: {}".format(new_height))
    if new_height == last_height:
        break
    last_height = new_height
    
    cnt += 1
```
